```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("/content/employees_data.csv")
print("EMPLOYEE DATA")
print(df.head())

#  Dataset Information
print("Dataset Information")
print(df.info())
print("Dataset Shape")
print(df.shape)
print("Column Names")
print(df.columns)

#  Missing Values
print("Missing Values")
print(df.isnull().sum())

# Fill missing Age with average age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing Gender with Mode
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

# Fill missing Salary with average salary
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
print("Statistics")
print(df.describe())

# Unique Job Titles
print("Job Titles")
print(df["JobTitle"].unique())

# Employee Count by Gender
print("Employees by Gender")
print(df["Gender"].value_counts())

# Employee Count by Job Title
print("Employees by Job Title")
print(df["JobTitle"].value_counts())

#  Average Salary by Job Title
print("Average Salary by Job Title")
print(df.groupby("JobTitle")["Salary"].mean())


#  Average Salary by Gender
print("Average Salary by Gender")
print(df.groupby("Gender")["Salary"].mean())


# Step 10 : Average Salary by Location
print("Average Salary by Location")
print(df.groupby("Location")["Salary"].mean())


#  Correlation
print("Correlation")
print(df[["Age", "Salary"]].corr())

# Salary Distribution
plt.figure(figsize=(7,5))
plt.hist(df["Salary"], bins=10)
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Employees")
plt.show()


# Gender Distribution
plt.figure(figsize=(6,4))
sns.countplot(x="Gender", data=df)
plt.title("Gender Distribution")
plt.show()


# Age Distribution
plt.figure(figsize=(7,5))
plt.hist(df["Age"], bins=10)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Employees")
plt.show()

# Average Salary by Job Title
plt.figure(figsize=(10,5))
sns.barplot(x="JobTitle", y="Salary", data=df)
plt.xticks(rotation=45)
plt.title("Average Salary by Job Title")
plt.show()


# Salary by Gender
plt.figure(figsize=(6,4))
sns.boxplot(x="Gender", y="Salary", data=df)
plt.title("Salary by Gender")
plt.show()


# Salary by Location
plt.figure(figsize=(8,5))
sns.barplot(x="Location", y="Salary", data=df)
plt.xticks(rotation=45)
plt.title("Average Salary by Location")
plt.show()


# Age vs Salary
plt.figure(figsize=(7,5))
sns.scatterplot(x="Age", y="Salary", data=df)
plt.title("Age vs Salary")
plt.show()

# Correlation Heatmap
plt.figure(figsize=(5,4))
sns.heatmap(df[["Age", "Salary"]].corr(),
            annot=True,
            cmap="Blues")
plt.title("Correlation Heatmap")
plt.show()

# Insights
print("========== INSIGHTS ==========")

print(" Highest average salary by Job Title:")
print(df.groupby("JobTitle")["Salary"].mean().idxmax())

print(" Average salary of Male and Female:")
print(df.groupby("Gender")["Salary"].mean())

print("Highest paying Location:")
print(df.groupby("Location")["Salary"].mean().idxmax())

print(" Total Employees:")
print(len(df))

print(" Average Employee Age:")
print(round(df["Age"].mean(),2))

print("Analysis Completed Successfully")
```
